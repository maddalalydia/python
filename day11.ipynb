Types of argument:
* Positional arg
* Keyword arg
* Default arg
* Variable length positional arg
* Variable length keyword arg

```python
# sub of 2 numbers
def sub(n):
    total = 0
    for i in n:
        total = i - total
    return total
num = [20, 30]
print("Sum is:", sub(num))
```

```python
#sub of 2 num
def sub(x,y):
    return x-y
b1 = int(input())
b2 = int(input())
print(sub(b1,b2))
print(sub(b2,b1))
```

```python
#Keyword arg
#sub of 2 num(it gives +ve output)
def sub(x,y):
    return x-y
b1 = int(input())
b2 = int(input())
print(sub(x=b1,y=b2))
print(sub(y=b2,x=b1))
```

```python
#sum of 2 num
def sum(x,y):
    return x+y
b1 = int(input())
b2 = int(input())
print(sum(b1,b2))
```

```python
def sum(a,b,c):
    return a+b+c
a = int(input())
b = int(input())
c = int(input())
print(sum(a,b,c))
```

```python
def sum(a,b,c,d):
    return a+b+c
a = int(input())
b = int(input())
c = int(input())
d = int(input())
print(sum(a,b,c,d))
```

```python
def sum(a,b,c=0,d=0):
    return a+b+c+d
a = int(input())
b = int(input())
c = int(input())
d = int(input())
print(sum(a,b))
print(sum(a,b,c))
print(sum(a,b,c,d))
```

```python
def mul(a, b, c=1, d=1):
    return a * b * c * d
a = int(input())
b = int(input())
c = int(input())
d = int(input())
print(mul(a, b))
print(mul(a, b, c))
print(mul(a, b, c, d))
```

LIST:
   List is a collection of different data type elements and it is mutable, List is defined with '[ ]'
Characterstics:
* list is ordered collection
* list is mutable
* elements can be accessed by using index
* list is sliceable
* list are hetrogeneous data type
* list is unhashable



```python
l1=[1,2,3,4]
l2=[5,7,8,7,9]
l3=l1+l2
print(l3)
```

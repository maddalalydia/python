# Removing elements from list
* remove(val) : It removes the first accurance value from list,otherwise value error
* pop(pos) : It remove and return value from list, if the list is empty it raises index error
* pop(position) : It remove and return specific position element from list
* clear() : it removes all elements from list

```python
l = [1,2,3,10,4,5]
print(l)
l.remove(10)
print(l)
l.pop(3)
print(l)
val = l.pop() #pops the last element 
print(val)
print(l)
print(l.pop(),l) # print last element within the print stmt
##rint(l.pop(120),l)   #pop index out of range
l.clear()
print(l)
#del l
#print(l)  #name 'l' is not defined
```

```python
# index(ele) : It returns the index of the first occurance value, if element not in the list
l = [1,2,3,4,5,2,4,6,6,7]
val = l.index(4)
print(val)
print(l.index(5))
print(l.index(4,5))
print(l)
#print(l.index(100))  # list.index(x): x not in list
```

```python
# Count(ele) : It returns the total occurance of an ele in a list
l = [1,2,3,4,3,4,3,4]
print(l.count(4))
print(l.count(5))
```

```python
total = int(input("Enter total number of students: "))
present = 0
absent = 0
i = 1
while i <= total:
    attendance = int(input("Enter attendance for student(1=Present, 0=Absent): "))
    if attendance == 1:
        present += 1
        i += 1
    elif attendance == 0:
        absent += 1
        i += 1
    else:
        print("Invalid input! Please enter 1 or 0 for the same student.")
print("Total Present:", present)
print("Total Absent:", absent)
percentage = int((present / total) * 100)
print("Attendance Percentage:", percentage, "%")
```

```python
password = input()
lower_ind = upper_ind = digit_ind = special_ind = False
for char in password:
    if char.isupper():
        upper_ind=True
    elif char.islower():
        lower_ind=True
    elif char.isdigit():
        digit_ind=True
    elif char in '@#$.!^':
        special_ind=True
if lower_ind and upper_ind and digit_ind and special_ind:
    print("Strong password")
else:
    print("Weak password")
```

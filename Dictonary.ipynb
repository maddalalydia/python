# Dictonary:
  dictonary is ordered collection of key-variable pairs. It is defined with '{ }'
# Charactesrstics: ex: {key:value}, {'Name':'Lydia'}
  * Dictonary keys are unique.
  * Dictonary values allows duplicate.
  * Dictonary values can access by using keys.
  * Dictonary dosent supports index and slicing.
  * list can't be used as dictonary key.
  * dictonary is mutable.
  * it can change the positions and values.
  * Key and value is seperated by using ':'.
  

```python
# dict is an ordered collection
details = {'name':'Lydia','batch':49,'course':'python'}
print(details)
```

```python
# if any duplicate value is there it replaces with the last value
details = {'name':'Lydia','batch':49,'course':'python','name':'Kiran'}
print(details)
```

```python
# to return specific value
details = {'name':'Lydia','batch':49,'course':'python','name':'Kiran'}
print(details['name'])
```

```python
# to return error if the value  not present in the dict
details = {'name':'Lydia','batch':49,'course':'python','name':'Kiran'}
print(details [0])    # KeyError: 0
```

```python
# to return specific value
details = {'name':'Lydia','batch':49,'course':'python','name':'Kiran',0:1}
print(details [0])
```

```python
# dictonary is mutable it can add or changes the val
details = {'name':'Lydia','batch':49,'course':'python','name':'Lydia Kiran'}
details['course'] = 'java'
details['age'] = 21
print(details)
```

```python
# dictonary methods
# get(key,default_val): it returns the value if the key is present in the dictonary otherwise it returns default value
details = {'name':'Lydia','batch':49,'course':'python','name':'Kiran'}
print(details.get('location','india'))
print(details.get('name','ram'))
print(details.get('marks'))
```

```python
# update(item): Dict updates with new items
details = {'name':'Lydia','batch':49,'course':'python','name':'Kiran',0:1}
new_dict = {'course':'java','batch':'50','location':'india'}
details.update(new_dict)
print(details)
```

```python
# Keys(): It returns the list of dict keys.
# values(): It returns the list of dict values.
# items(): it returns the dict list of tuple of key-value pairs.
details = {'name':'Lydia','batch':49,'course':'python','name':'Kiran',0:1}
new_dict = {'course':'java','batch':'50','location':'india'}
details.update(new_dict)
print(details)
print(details.keys())
print(details.values())
print(details.items())
print(list(details.items()))
```

```python
#pop(key): it returns the value and remove the item from dict if thr key is present,otherwise key error.
#popitem(): it returns and removes the last item.
details = {'name':'Lydia','batch':49,'course':'python','name':'Kiran',0:1}
print(details.pop('name'))
print(details.popitem())
print(details)
```

```python
#op overloading
class Math:
    def __init__(self, value):
        self.value = value
    def __add__(self, other):
        return Math(self.value + other.value)  
    def __str__(self):
        return str(self.value)
m1 = Math(10)
m2 = Math(5)
m3 = Math(7)
m4 = Math(8)
print(m1 + m2)        
print(m2 + m3)        
print(m2 + m3 + m4)
```

```python
#Duck type
class AddTwo:
    def __init__(self, a, b):
        self.a = a
        self.b = b
    def calculate(self):
        return self.a + self.b
class AddThree:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    def calculate(self):
        return self.a + self.b + self.c
class AddFour:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d
    def calculate(self):
        return self.a + self.b + self.c + self.d
def result(obj):
    print(obj.calculate()) 
a1 = AddTwo(10, 5)
a2 = AddThree(5, 6, 7)
a3 = AddFour(5, 6, 7, 8)
result(a1)
result(a2)
result(a3)
```

# set operations or set methods
* Union(new_set) :  It returns the all unique elements from set.
* Intersection(new_set) :  it returns only common elements in a set. 
* Difference(difference) :  retuens only the elements which are not present in the other set.
* Symmetric difference(symmetric_difference) :  return only non common elements.
* subset(issubset) :  Set of all elements fresent in the other set.
* Superset(issuperset) :  Reverse of subset.
* Disjoint set(isdisjoint) :  where a set is not present in any other set.

```python
s1 = {1,2,3,6}
s2 = {3,4,5,6}
print(s.union(s2))
print(s1 | s2)
print("intersection")
print(s1.intersection(s2))
print(s1 & s2)
print("Difference")
print(s1.difference(s2))
print(s1-s2)
print("Symmetric difference")
print(s1.symmetric_difference(s2))
print("Subset")
print(s1.issubset(s2))
print({1,2}.issubset(s1))
print({1,2,5}.issubset(s1))
print(set().issubset(s1))  # every empty set is a subset
# Super set
print("Super set***********")
print(s2.issuperset(s1))
print(s1.issuperset({1,2}))
print({1,2,5}.issuperset(s1))
print("Disjoint set************")
print(s1.isdisjoint(s2))
print({1,2}.isdisjoint({10}))
```

# add(ele) : 
* It add to set , if the elements is not present in set, otherwise nothing.
# update(new_set) : 
* Set elements update with a new set elements .
# remove(ele) : 
* It removes the set element, if element not present in set it raises key error.
# discard(ele) : 
* It removes the ele from set otherwise nothing.

```python
s = {1,2,3,4}
s.add(10)
print(s)
s.update({1,2,5,6})
print(s)
s.remove(10)
print(s)
# remove(20) Key error
s.discard(100)
print(s)
s.discard(5)
print(s)
```

```python
print(min(s))
print(max(s))
```

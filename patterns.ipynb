```python
# Seperate even and odd numbers from list og numbers
l = [1, 5, 6, 7, 9, 10, 12]
even = []
odd = []
for i in l:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)
print("Even numbers:", even)
print("Odd numbers:", odd)
```

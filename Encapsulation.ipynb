```python
class Bank:
    bank_name = "My National Bank"   # class attribute
    total_users = 0
    total_bank_balance = 0
    users_list = []   
    def __init__(self, name, mail, password, amount):
        self.name = name
        self.mail = mail
        self.password = password
        self.balance = amount
        Bank.total_users += 1
        Bank.total_bank_balance += amount
        Bank.users_list.append(self)
    def deposit(self, amount):
        self.balance += amount
        Bank.total_bank_balance += amount
        print(f"{amount} deposited successfully")
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            Bank.total_bank_balance -= amount
            print(f"{amount} withdrawn successfully")
        else:
            print("Insufficient balance")
    def check_balance(self):
        print(f"Available balance: {self.balance}")
    @classmethod
    def total_users_info(cls):
        return f"Total users in bank is {cls.total_users}"
    @classmethod
    def total_bank_amount(cls):
        return f"Total amount in bank is {cls.total_bank_balance}"
    @classmethod
    def login(cls, mail, password):
        for user in cls.users_list:
            if user.mail == mail and user.password == password:
                return user
        return None
while True:
    print("\n--- Welcome to Bank System ---")
    print("1. Create Account")
    print("2. Login")
    print("3. Bank Details")
    print("4. Exit")
    choice = input("Enter choice: ")
    if choice == "1":
        name = input("Enter name: ")
        mail = input("Enter mail: ")
        password = input("Enter password: ")
        amount = float(input("Enter initial amount: "))
        Bank(name, mail, password, amount)
        print("Account created successfully!")
    elif choice == "2":
        mail = input("Enter mail: ")
        password = input("Enter password: ")
        user = Bank.login(mail, password)
        if user:
            print(f"\nWelcome {user.name}!")
            while True:
                print("\n1. Deposit")
                print("2. Withdraw")
                print("3. Check Balance")
                print("4. Logout")
                op = input("Enter option: ")
                if op == "1":
                    amt = float(input("Enter amount: "))
                    user.deposit(amt)
                elif op == "2":
                    amt = float(input("Enter amount: "))
                    user.withdraw(amt)
                elif op == "3":
                    user.check_balance()
                elif op == "4":
                    print("Logged out")
                    break
                else:
                    print("Invalid option")
        else:
            print("Invalid login details")
    elif choice == "3":
        print(Bank.total_users_info())
        print(Bank.total_bank_amount)
    elif choice == "4":
        print("Thank you!")
        break
    else:
        print("Invalid choice")
```

```python
n = 10
present = 0
absent = 0

for i in range(1, n + 1):
    val = int(input(f'Enter Roll no {i} attendance (1=Present, 0=Absent): '))
    
    if val == 1:
        present += 1
    else:
        absent += 1

percentage = (present / n) * 100

print("Total Present:", present)
print("Total Absent:", absent)
print("Attendance Percentage:", percentage, "%")
```

```python
# Method overriding
class Employee:
    def __init__(self, name, id, sal):
        self.name = name
        self.id = id
        self.sal = sal
    def show_details(self):
        return f"My name is {self.name}, ID is {self.id} and salary is {self.sal}"
class Student(Employee):
    def __init__(self, name, id, sal, course):
        super().__init__(name, id, sal)
        self.course = course
    def show_details(self):
        parent_details = super().show_details()
        return f"{parent_details} and I am studying {self.course}"
e = Employee("Dia", 101, 30000)
s = Student("Kiran", 201, 0, "B.Tech")
print(s.show_details())
print(e.show_details())
print(Employee.__mro__)     #Method resolution order
```

```python
#method overloading
class math():
    def add(self,x,y):
        return x+y  
    def add(self,x,y,z):
        return x+y+z
    def add(self,a,b,c = 0,d = 0):
        return a+b+c+d
m = math()
print(m.add(10,5))    
print(m.add(5,6,7))
print(m.add(5,6,7,8))
```

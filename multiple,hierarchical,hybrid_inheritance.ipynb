```python
# Multiple
# Parent class 1
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def show_details(self):
        return f"My name is {self.name} and I'm {self.age} years old"
class Department:
    def __init__(self, dept):
        self.dept = dept
    def show_dept(self):
        return f"I am a student at {self.dept}"
class Employee(Person, Department):
    def __init__(self, emid, name, age, role, dept):
        Person.__init__(self, name, age)
        Department.__init__(self, dept)
        self.emid = emid
        self.role = role
    def show_details(self):
        print(super().show_details(), end=" ")
        print(self.show_dept(), end=" ")
        return f"and my role is {self.role}."
e = Employee(101, "Lydia", 21, "Student", "Codegnan")
print(e.show_details())
```

```python
#hierarchical
# Parent class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def show_details(self):
        return f"My name is {self.name} and I'm {self.age} years old"
class Student(Person):
    def __init__(self, name, age, institute):
        super().__init__(name, age)
        self.institute = institute
    def show_student(self):
        return f"I am a student at {self.institute}"
class Employee(Person):
    def __init__(self, name, age, job):
        super().__init__(name, age)
        self.job = job
    def show_employee(self):
        return f"I am working as a {self.job}"
s = Student("Lydia", 21, "Codegnan")
e = Employee("Lydia", 21, "Trainer")
print(s.show_details(), s.show_student())
print(e.show_details(), e.show_employee())
```

```python
# Hybrid
class Person:
    def __init__(self, name):
        self.name = name
    def show_name(self):
        return f"My name is {self.name}"
class Student(Person):
    def study(self):
        return "I am studying"
class Employee(Person):
    def work(self):
        return "I am working"
class Intern(Student, Employee):
    def role(self):
        return "I am a student intern"
i = Intern("Lydia")
print(i.show_name())
print(i.study())
print(i.work())
print(i.role())
```
